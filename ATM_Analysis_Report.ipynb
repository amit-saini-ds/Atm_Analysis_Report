#*Q11. Calculate total Monthly_Transactions city-wise.*

```python
monthly_transactions =df.groupby("City")["Monthly_Transactions"].sum()

display(monthly_transactions)
```

```python
# Represent by Graph – City-wise Total Monthly Transactions

plt.bar(monthly_transactions.index,monthly_transactions.values,color = ["dodgerblue","firebrick","dodgerblue","dodgerblue"])
plt.xlabel("City")
plt.ylabel("Total Transactions")
plt.title("City-wise Total Monthly Transactions")

plt.show()
```

## *Q29. Calculate cash utilization percentage for each city.*

```python
percentage_utilization = df.groupby("City")[["Cash_Withdrawn_Lakhs","Cash_Loaded_Lakhs"]].sum()
percentage_utilization["percentage_utilization"] = (percentage_utilization["Cash_Withdrawn_Lakhs"]/percentage_utilization["Cash_Loaded_Lakhs"])*100

display(percentage_utilization)
```

```python
# Represent by Graph - Cash utilization percentage for each city

plt.bar(percentage_utilization["Cash_Withdrawn_Lakhs"],labels = percentage_utilization.index,autopct = '%1.2f%%')
plt.title("Cash utilization percentage for each city")

plt.show()
```
